```python
# 6. Выравнивание прочтений и преобразование в BAM

# Удаляем старые временные/битые файлы перед запуском, если они остались от прерываний
!rm -f aligned_sorted.bam aligned_sorted.bam.bai aligned_sorted.bam.tmp.*

# пайплайн без промежуточного SAM-файла — быстрее (ненамного)
!bwa mem -t 2 ecoli_k12.fasta trimmed_1P.fastq trimmed_2P.fastq | \
  samtools view -bS - | \
  samtools sort -@ 2 -o aligned_sorted.bam

# Индексация
!samtools index aligned_sorted.bam
```

```python
# 7. Поиск SNP и создание VCF-файла

# Используем bcftools mpileup вместо устаревшего samtools mpileup
!bcftools mpileup -Ou -f ecoli_k12.fasta aligned_sorted.bam | \
  bcftools call -mv --ploidy 1 -Oz -o variants.vcf.gz

# Индексация VCF-файла
!bcftools index variants.vcf.gz

# Просмотр первых 20 вариантов
!bcftools view variants.vcf.gz | head -n 20
```

```python

# =============================================================
# 🧬 Блок 8. Аннотация SNP и поиск генов (через NCBI Entrez)
# =============================================================
from Bio import Entrez
Entrez.email = "your_email@example.com"  # Замените на вашу почту
# Автоматическое извлечение первых 5 координат SNP из VCF
import gzip

variant_positions = []

with gzip.open("variants.vcf.gz", "rt") as f:
    for line in f:
        if line.startswith("#"):
            continue
        fields = line.strip().split("	")
        chrom = fields[0]
        pos = fields[1]
        variant_positions.append(f"{chrom}:{pos}")
        if len(variant_positions) == 5:
            break

# Автоматическая аннотация через NCBI Entrez efetch
from time import sleep
from Bio import SeqIO

for vp in variant_positions:
    chrom, pos = vp.split(":")
    print(f"\nПроверка {chrom}:{pos}...")
    try:
        handle = Entrez.efetch(db="nucleotide", id="NC_000913.3", rettype="gb", strand=1,
                               seq_start=int(pos), seq_stop=int(pos)+1)
        record = SeqIO.read(handle, "genbank")
        handle.close()

        genes = set()
        for feature in record.features:
            if feature.type in ["gene", "CDS"]:
                gene = feature.qualifiers.get("gene", ["неизвестно"])[0]
                genes.add(gene)

        if genes:
            for g in genes:
                print(f"  → Найден ген: {g}")
        else:
            print("  → Мутация вне известных генов")

    except Exception as e:
        print("  ⚠️ Ошибка при аннотации:", e)
    sleep(0.5)  # Пауза, чтобы не перегружать сервер
```

### Оценка мутаций

**1. Мутации в гене `talB` (позиции: 8723, 8948, 8960)**
- Ген `talB` кодирует транзальдолазу B, участвующую в пентозофосфатном пути (PPP).
- Этот путь критичен для образования NADPH (редокс-защита) и рибозы (синтез нуклеотидов).
- **Влияние мутации:**
  - Ослабление окислительной защиты
  - Нарушение метаболизма при действии меди
  - Возможное снижение устойчивости к антибиотикам, индуцирующим ROS

**2. Мутация в гене `carB` (позиция: 30834)**
- Ген `carB` участвует в синтезе карбамоилфосфата — ключевого метаболита для синтеза аргинина и пиримидинов.
- **Влияние мутации:**
  - Потенциальное снижение темпов роста
  - Нарушение синтеза белков и нуклеотидов
  - Повышенная чувствительность к антиметаболитам

**3. Мутация вне известных генов (позиция: 30814)**
- Не принадлежит к области известных CDS или генов
- **Возможное влияние:**
  - Может располагаться в регуляторной области (промотор, оператор)
  - Может не иметь функционального значения без дальнейшей проверки
